```python
# =================================================================================================
# CELL 5A — FINAL SURROGATE EXPLAINABILITY
# Tensile + Roughness
#
# Final Ridge-Stacking prediction function
# Exact/model-agnostic SHAP
# Held-out permutation importance
# PDP + ICE
# Two-way PDP interaction surfaces
# Q1-quality 600-DPI PNG + vector PDF
# =================================================================================================

import sys
import subprocess
import importlib.util
import warnings
import json
import gc
from pathlib import Path

warnings.filterwarnings("ignore")


# =================================================================================================
# 1. PACKAGE CHECK
# =================================================================================================

required = {
    "numpy": "numpy",
    "pandas": "pandas",
    "matplotlib": "matplotlib",
    "joblib": "joblib",
    "shap": "shap",
    "openpyxl": "openpyxl"
}

for module, package in required.items():

    if importlib.util.find_spec(module) is None:

        subprocess.check_call([
            sys.executable,
            "-m",
            "pip",
            "install",
            "-q",
            package
        ])


# =================================================================================================
# 2. IMPORTS
# =================================================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import joblib
import shap

from sklearn.metrics import mean_squared_error
from IPython.display import display


# =================================================================================================
# 3. SETTINGS
# =================================================================================================

SEED = 20260906

PERMUTATION_REPEATS = 30

SHAP_BACKGROUND_N = 40

SHAP_EXPLAIN_N = 120

ICE_SAMPLE_N = 120

PDP_GRID_N = 40

INTERACTION_BACKGROUND_N = 150

INTERACTION_GRID_N = 20


rng = np.random.default_rng(
    SEED
)


pd.set_option(
    "display.max_columns",
    100
)

pd.set_option(
    "display.width",
    220
)

pd.set_option(
    "display.float_format",
    lambda x: f"{x:.6f}"
)


plt.rcParams.update({
    "font.size": 11,
    "axes.titlesize": 12,
    "axes.labelsize": 11,
    "xtick.labelsize": 10,
    "ytick.labelsize": 10
})


# =================================================================================================
# 4. FIND PROJECT
# =================================================================================================

candidate_projects = [

    Path(
        r"C:\Users\apurb\Downloads\Simulation OP ABS"
    ),

    Path.cwd(),

    Path.home() /
    "Downloads" /
    "Simulation OP ABS"
]


PROJECT_ROOT = None


for root in candidate_projects:

    tensile_manifest = (
        root /
        "Q1_ML_Output" /
        "Cell4_ML_Benchmarking" /
        "Cell4B_Tensile_Deep_Tuning" /
        "Cell4B_Tensile_Manifest.json"
    )

    roughness_manifest = (
        root /
        "Q1_ML_Output" /
        "Cell4_ML_Benchmarking" /
        "Cell4B_Roughness_Deep_Tuning" /
        "Cell4B_Roughness_Manifest.json"
    )

    if (
        tensile_manifest.exists()
        and
        roughness_manifest.exists()
    ):

        PROJECT_ROOT = root.resolve()

        TENSILE_MANIFEST_FILE = (
            tensile_manifest.resolve()
        )

        ROUGHNESS_MANIFEST_FILE = (
            roughness_manifest.resolve()
        )

        break


if PROJECT_ROOT is None:

    raise FileNotFoundError(
        "Cell 4B Tensile/Roughness manifests were not found."
    )


# =================================================================================================
# 5. LOAD MANIFESTS
# =================================================================================================

with open(
    TENSILE_MANIFEST_FILE,
    "r",
    encoding="utf-8"
) as f:

    tensile_manifest = json.load(f)


with open(
    ROUGHNESS_MANIFEST_FILE,
    "r",
    encoding="utf-8"
) as f:

    roughness_manifest = json.load(f)


# =================================================================================================
# 6. LOAD FINAL MODEL BUNDLES
# =================================================================================================

TENSILE_MODEL_FILE = Path(
    tensile_manifest[
        "final_model_bundle"
    ]
)


ROUGHNESS_MODEL_FILE = Path(
    roughness_manifest[
        "final_model_bundle"
    ]
)


if not TENSILE_MODEL_FILE.exists():

    raise FileNotFoundError(
        TENSILE_MODEL_FILE
    )


if not ROUGHNESS_MODEL_FILE.exists():

    raise FileNotFoundError(
        ROUGHNESS_MODEL_FILE
    )


tensile_bundle = joblib.load(
    TENSILE_MODEL_FILE
)


roughness_bundle = joblib.load(
    ROUGHNESS_MODEL_FILE
)


print(
    "Tensile final model  :",
    tensile_bundle[
        "selected_model"
    ]
)


print(
    "Roughness final model:",
    roughness_bundle[
        "selected_model"
    ]
)


# =================================================================================================
# 7. LOAD CELL 3 DATA
# =================================================================================================

DATA_FILE = Path(
    tensile_manifest[
        "data_file"
    ]
)


data = pd.read_csv(
    DATA_FILE
)


INPUT_COLS = tensile_manifest[
    "input_cols"
]


TENSILE_TARGET = tensile_manifest[
    "target"
]


ROUGHNESS_TARGET = roughness_manifest[
    "target"
]


FEATURE_LABELS = [

    "Glass Fiber (%)",

    "Infill Density (%)",

    "Layer Thickness (mm)",

    "Nozzle Temperature (°C)",

    "Bed Temperature (°C)",

    "Printing Speed (mm/s)"
]


FEATURE_MAP = dict(
    zip(
        INPUT_COLS,
        FEATURE_LABELS
    )
)


X_ALL = data[
    INPUT_COLS
].to_numpy(
    dtype=float
)


y_tensile = data[
    TENSILE_TARGET
].to_numpy(
    dtype=float
)


y_roughness = data[
    ROUGHNESS_TARGET
].to_numpy(
    dtype=float
)


# =================================================================================================
# 8. FINAL BUNDLE PREDICTION FUNCTION
# =================================================================================================

def predict_bundle(
    bundle,
    X
):

    if isinstance(
        X,
        pd.DataFrame
    ):

        X = X.to_numpy(
            dtype=float
        )

    else:

        X = np.asarray(
            X,
            dtype=float
        )


    selected = bundle[
        "selected_model"
    ]


    base_models = bundle[
        "base_models"
    ]


    base_order = bundle[
        "base_model_order"
    ]


    # Selected single base model
    if selected in base_models:

        return np.asarray(
            base_models[
                selected
            ].predict(
                X
            )
        ).ravel()


    # Base prediction matrix
    P = np.column_stack([

        base_models[
            name
        ].predict(
            X
        )

        for name in base_order

    ])


    if (
        selected
        ==
        "Weighted_Ensemble"
    ):

        weights = np.asarray(
            bundle[
                "weighted_weights"
            ],
            dtype=float
        )

        return np.asarray(
            P @ weights
        ).ravel()


    if (
        selected
        ==
        "Ridge_Stacking"
    ):

        return np.asarray(

            bundle[
                "stacker"
            ].predict(
                P
            )

        ).ravel()


    raise ValueError(
        f"Unknown bundle model: {selected}"
    )


# =================================================================================================
# 9. VERIFY FINAL MODEL PREDICTION
# =================================================================================================

for label, bundle in [

    (
        "Tensile",
        tensile_bundle
    ),

    (
        "Roughness",
        roughness_bundle
    )

]:

    check_pred = predict_bundle(
        bundle,
        X_ALL[:10]
    )

    if (
        len(check_pred)
        !=
        10
    ):

        raise RuntimeError(
            f"{label} prediction wrapper failed."
        )


print(
    "\nFinal model prediction wrappers: OK"
)


# =================================================================================================
# 10. OUTPUT DIRECTORIES
# =================================================================================================

OUTPUT_DIR = (

    PROJECT_ROOT /
    "Q1_ML_Output" /
    "Cell5_XAI_Sensitivity" /
    "Cell5A_Explainability"

)


TABLE_DIR = (
    OUTPUT_DIR /
    "Tables"
)


FIGURE_DIR = (
    OUTPUT_DIR /
    "Figures"
)


SHAP_DIR = (
    OUTPUT_DIR /
    "SHAP"
)


PERM_DIR = (
    OUTPUT_DIR /
    "Permutation_Importance"
)


PDP_DIR = (
    OUTPUT_DIR /
    "PDP_ICE"
)


INTERACTION_DIR = (
    OUTPUT_DIR /
    "Interactions"
)


for folder in [

    OUTPUT_DIR,

    TABLE_DIR,

    FIGURE_DIR,

    SHAP_DIR,

    PERM_DIR,

    PDP_DIR,

    INTERACTION_DIR

]:

    folder.mkdir(
        parents=True,
        exist_ok=True
    )


# =================================================================================================
# 11. FIXED TEST SET FOR POST-HOC PERMUTATION IMPORTANCE
#
# Important:
# Models are already frozen.
# This section does NOT tune or select models.
# =================================================================================================

TEST_IDX_T = np.asarray(
    tensile_bundle[
        "test_indices"
    ],
    dtype=int
)


TEST_IDX_R = np.asarray(
    roughness_bundle[
        "test_indices"
    ],
    dtype=int
)


if not np.array_equal(
    TEST_IDX_T,
    TEST_IDX_R
):

    raise RuntimeError(
        "Tensile and Roughness test indices are not identical."
    )


TEST_IDX = TEST_IDX_T


X_TEST = X_ALL[
    TEST_IDX
]


Y_TEST_T = y_tensile[
    TEST_IDX
]


Y_TEST_R = y_roughness[
    TEST_IDX
]


# =================================================================================================
# 12. MANUAL MODEL-AGNOSTIC PERMUTATION IMPORTANCE
#
# Importance = increase in RMSE after shuffling feature.
# =================================================================================================

def permutation_rmse_importance(
    bundle,
    X,
    y,
    response_name,
    repeats=30
):

    baseline_pred = predict_bundle(
        bundle,
        X
    )


    baseline_rmse = np.sqrt(

        mean_squared_error(
            y,
            baseline_pred
        )

    )


    rows = []


    local_rng = np.random.default_rng(
        SEED
    )


    for j, feature in enumerate(
        INPUT_COLS
    ):

        increases = []


        for repeat in range(
            repeats
        ):

            X_perm = X.copy()


            X_perm[
                :,
                j
            ] = local_rng.permutation(

                X_perm[
                    :,
                    j
                ]

            )


            pred = predict_bundle(

                bundle,

                X_perm

            )


            perm_rmse = np.sqrt(

                mean_squared_error(
                    y,
                    pred
                )

            )


            increases.append(

                perm_rmse -
                baseline_rmse

            )


        rows.append({

            "Response":
                response_name,

            "Feature":
                feature,

            "Feature_Label":
                FEATURE_MAP[
                    feature
                ],

            "Baseline_RMSE":
                baseline_rmse,

            "RMSE_Increase_Mean":
                np.mean(
                    increases
                ),

            "RMSE_Increase_SD":
                np.std(
                    increases,
                    ddof=1
                ),

            "RMSE_Increase_Min":
                np.min(
                    increases
                ),

            "RMSE_Increase_Max":
                np.max(
                    increases
                )
        })


    result = (

        pd.DataFrame(
            rows
        )

        .sort_values(
            "RMSE_Increase_Mean",
            ascending=False
        )

        .reset_index(
            drop=True
        )

    )


    result[
        "Permutation_Rank"
    ] = np.arange(

        1,

        len(
            result
        ) + 1

    )


    return result


print(
    "\nRunning final-model permutation importance..."
)


perm_t = permutation_rmse_importance(

    tensile_bundle,

    X_TEST,

    Y_TEST_T,

    "Tensile Strength",

    repeats=
        PERMUTATION_REPEATS

)


perm_r = permutation_rmse_importance(

    roughness_bundle,

    X_TEST,

    Y_TEST_R,

    "Surface Roughness",

    repeats=
        PERMUTATION_REPEATS

)


perm_t.to_csv(

    PERM_DIR /
    "Tensile_Final_Model_Permutation_Importance.csv",

    index=False

)


perm_r.to_csv(

    PERM_DIR /
    "Roughness_Final_Model_Permutation_Importance.csv",

    index=False

)


print(
    "\nTENSILE — PERMUTATION IMPORTANCE"
)

display(
    perm_t.round(
        6
    )
)


print(
    "\nROUGHNESS — PERMUTATION IMPORTANCE"
)

display(
    perm_r.round(
        6
    )
)


# =================================================================================================
# 13. PERMUTATION FIGURE
# =================================================================================================

def plot_permutation(
    table,
    response_name,
    unit,
    filename
):

    temp = table.sort_values(
        "RMSE_Increase_Mean",
        ascending=True
    )


    plt.figure(
        figsize=(
            8,
            5.5
        )
    )


    plt.barh(

        temp[
            "Feature_Label"
        ],

        temp[
            "RMSE_Increase_Mean"
        ],

        xerr=
            temp[
                "RMSE_Increase_SD"
            ]

    )


    plt.xlabel(
        f"Increase in RMSE ({unit})"
    )


    plt.ylabel(
        "Input Parameter"
    )


    plt.title(
        f"{response_name} — Permutation Importance"
    )


    plt.tight_layout()


    plt.savefig(

        PERM_DIR /
        f"{filename}.png",

        dpi=
            600,

        bbox_inches=
            "tight"

    )


    plt.savefig(

        PERM_DIR /
        f"{filename}.pdf",

        bbox_inches=
            "tight"

    )


    plt.show()


plot_permutation(

    perm_t,

    "Tensile Strength",

    "MPa",

    "Tensile_Permutation_Importance"

)


plot_permutation(

    perm_r,

    "Surface Roughness",

    "µm",

    "Roughness_Permutation_Importance"

)


# =================================================================================================
# 14. REPRESENTATIVE SHAP SAMPLES
# =================================================================================================

all_random_indices = rng.permutation(
    len(
        X_ALL
    )
)


background_idx = all_random_indices[
    :SHAP_BACKGROUND_N
]


explain_idx = all_random_indices[

    SHAP_BACKGROUND_N:

    SHAP_BACKGROUND_N +
    SHAP_EXPLAIN_N

]


X_BACKGROUND = pd.DataFrame(

    X_ALL[
        background_idx
    ],

    columns=
        FEATURE_LABELS

)


X_EXPLAIN = pd.DataFrame(

    X_ALL[
        explain_idx
    ],

    columns=
        FEATURE_LABELS

)


# =================================================================================================
# 15. MODEL-AGNOSTIC SHAP
#
# Only six inputs:
# Try Exact SHAP first.
# Fallback = permutation SHAP.
# =================================================================================================

def run_shap(
    bundle,
    response_name,
    prefix
):

    print(
        f"\nRunning SHAP — {response_name}"
    )


    predict_function = lambda X: predict_bundle(
        bundle,
        X
    )


    masker = shap.maskers.Independent(

        X_BACKGROUND,

        max_samples=
            SHAP_BACKGROUND_N

    )


    shap_method = (
        "Exact"
    )


    try:

        explainer = shap.Explainer(

            predict_function,

            masker=

                masker,

            algorithm=
                "exact",

            feature_names=
                FEATURE_LABELS,

            seed=
                SEED

        )


        shap_values = explainer(
            X_EXPLAIN
        )


    except Exception as exact_error:

        print(
            "   Exact SHAP fallback to permutation SHAP."
        )


        shap_method = (
            "Permutation"
        )


        explainer = shap.Explainer(

            predict_function,

            masker=
                masker,

            algorithm=
                "permutation",

            feature_names=
                FEATURE_LABELS,

            seed=
                SEED

        )


        shap_values = explainer(

            X_EXPLAIN,

            max_evals=
                65

        )


    # ---------------------------------------------------------------------------------------------
    # SHAP VALUES TABLE
    # ---------------------------------------------------------------------------------------------

    shap_matrix = np.asarray(
        shap_values.values
    )


    shap_df = pd.DataFrame(

        shap_matrix,

        columns=[
            f"SHAP_{name}"
            for name in FEATURE_LABELS
        ]

    )


    shap_df.insert(

        0,

        "Original_Row_Index",

        explain_idx

    )


    shap_df.to_csv(

        SHAP_DIR /
        f"{prefix}_SHAP_Values.csv",

        index=False

    )


    # ---------------------------------------------------------------------------------------------
    # GLOBAL MEAN ABSOLUTE SHAP
    # ---------------------------------------------------------------------------------------------

    mean_abs = np.mean(

        np.abs(
            shap_matrix
        ),

        axis=0

    )


    global_df = pd.DataFrame({

        "Response":
            response_name,

        "Feature":
            INPUT_COLS,

        "Feature_Label":
            FEATURE_LABELS,

        "Mean_Absolute_SHAP":
            mean_abs

    })


    global_df = (

        global_df

        .sort_values(
            "Mean_Absolute_SHAP",
            ascending=False
        )

        .reset_index(
            drop=True
        )

    )


    global_df[
        "SHAP_Rank"
    ] = np.arange(

        1,

        len(
            global_df
        ) + 1

    )


    global_df[
        "SHAP_Method"
    ] = shap_method


    global_df.to_csv(

        SHAP_DIR /
        f"{prefix}_Global_SHAP_Importance.csv",

        index=False

    )


    # ---------------------------------------------------------------------------------------------
    # SHAP BEESWARM
    # ---------------------------------------------------------------------------------------------

    shap.plots.beeswarm(

        shap_values,

        max_display=
            6,

        show=
            False

    )


    plt.title(
        f"{response_name} — SHAP Summary"
    )


    plt.tight_layout()


    plt.savefig(

        SHAP_DIR /
        f"{prefix}_SHAP_Beeswarm.png",

        dpi=
            600,

        bbox_inches=
            "tight"

    )


    plt.savefig(

        SHAP_DIR /
        f"{prefix}_SHAP_Beeswarm.pdf",

        bbox_inches=
            "tight"

    )


    plt.show()


    # ---------------------------------------------------------------------------------------------
    # SHAP BAR
    # ---------------------------------------------------------------------------------------------

    shap.plots.bar(

        shap_values,

        max_display=
            6,

        show=
            False

    )


    plt.title(
        f"{response_name} — Mean |SHAP|"
    )


    plt.tight_layout()


    plt.savefig(

        SHAP_DIR /
        f"{prefix}_SHAP_Global_Bar.png",

        dpi=
            600,

        bbox_inches=
            "tight"

    )


    plt.savefig(

        SHAP_DIR /
        f"{prefix}_SHAP_Global_Bar.pdf",

        bbox_inches=
            "tight"

    )


    plt.show()


    return (
        global_df,
        shap_method
    )


shap_t, shap_method_t = run_shap(

    tensile_bundle,

    "Tensile Strength",

    "Tensile"

)


shap_r, shap_method_r = run_shap(

    roughness_bundle,

    "Surface Roughness",

    "Roughness"

)


print(
    "\nTENSILE — GLOBAL SHAP"
)

display(
    shap_t.round(
        6
    )
)


print(
    "\nROUGHNESS — GLOBAL SHAP"
)

display(
    shap_r.round(
        6
    )
)


# =================================================================================================
# 16. COMBINED GLOBAL IMPORTANCE RANKING
#
# Permutation + SHAP ranks
# Lower combined rank = stronger overall importance
# =================================================================================================

def combined_ranking(
    perm_df,
    shap_df,
    response
):

    merged = perm_df[[
        "Feature",
        "Feature_Label",
        "Permutation_Rank",
        "RMSE_Increase_Mean"
    ]].merge(

        shap_df[[
            "Feature",
            "SHAP_Rank",
            "Mean_Absolute_SHAP"
        ]],

        on=
            "Feature",

        how=
            "inner"

    )


    merged[
        "Combined_Mean_Rank"
    ] = (

        merged[
            "Permutation_Rank"
        ]

        +

        merged[
            "SHAP_Rank"
        ]

    ) / 2


    merged = (

        merged

        .sort_values(
            "Combined_Mean_Rank",
            ascending=True
        )

        .reset_index(
            drop=True
        )

    )


    merged[
        "Combined_Rank"
    ] = np.arange(

        1,

        len(
            merged
        ) + 1

    )


    merged.insert(

        0,

        "Response",

        response

    )


    return merged


combined_t = combined_ranking(

    perm_t,

    shap_t,

    "Tensile Strength"

)


combined_r = combined_ranking(

    perm_r,

    shap_r,

    "Surface Roughness"

)


combined_t.to_csv(

    TABLE_DIR /
    "Tensile_Combined_XAI_Ranking.csv",

    index=False

)


combined_r.to_csv(

    TABLE_DIR /
    "Roughness_Combined_XAI_Ranking.csv",

    index=False

)


print(
    "\nTENSILE — COMBINED XAI RANKING"
)

display(
    combined_t.round(
        6
    )
)


print(
    "\nROUGHNESS — COMBINED XAI RANKING"
)

display(
    combined_r.round(
        6
    )
)


# =================================================================================================
# 17. PDP + ICE
#
# Manual implementation so it works directly with final stacking model.
# =================================================================================================

ice_indices = rng.choice(

    len(
        X_ALL
    ),

    size=min(
        ICE_SAMPLE_N,
        len(
            X_ALL
        )
    ),

    replace=False

)


X_ICE_BASE = X_ALL[
    ice_indices
].copy()


def calculate_pdp_ice(
    bundle,
    feature_index
):

    feature_values = X_ALL[
        :,
        feature_index
    ]


    grid = np.linspace(

        np.min(
            feature_values
        ),

        np.max(
            feature_values
        ),

        PDP_GRID_N

    )


    ice_matrix = np.zeros(

        (
            len(
                X_ICE_BASE
            ),
            len(
                grid
            )
        ),

        dtype=float

    )


    for g, value in enumerate(
        grid
    ):

        X_temp = X_ICE_BASE.copy()


        X_temp[
            :,
            feature_index
        ] = value


        ice_matrix[
            :,
            g
        ] = predict_bundle(

            bundle,

            X_temp

        )


    pdp = np.mean(

        ice_matrix,

        axis=0

    )


    return (
        grid,
        ice_matrix,
        pdp
    )


def run_pdp_ice(
    bundle,
    response_name,
    prefix,
    unit
):

    pdp_summary = []


    print(
        f"\nRunning PDP + ICE — {response_name}"
    )


    for j, feature in enumerate(
        INPUT_COLS
    ):

        grid, ice_matrix, pdp = (
            calculate_pdp_ice(
                bundle,
                j
            )
        )


        temp = pd.DataFrame({

            "Grid_Value":
                grid,

            "PDP_Prediction":
                pdp

        })


        temp.to_csv(

            PDP_DIR /
            f"{prefix}_{feature}_PDP.csv",

            index=False

        )


        ice_df = pd.DataFrame(

            ice_matrix.T,

            columns=[
                f"ICE_{i+1}"
                for i
                in range(
                    ice_matrix.shape[
                        0
                    ]
                )
            ]

        )


        ice_df.insert(

            0,

            "Grid_Value",

            grid

        )


        ice_df.to_csv(

            PDP_DIR /
            f"{prefix}_{feature}_ICE.csv",

            index=False

        )


        # -----------------------------------------------------------------------------------------
        # Figure
        # -----------------------------------------------------------------------------------------

        fig, ax = plt.subplots(

            figsize=(
                7,
                5.5
            )

        )


        for row in ice_matrix:

            ax.plot(

                grid,

                row,

                alpha=
                    0.08,

                linewidth=
                    0.6

            )


        ax.plot(

            grid,

            pdp,

            linewidth=
                3.0,

            label=
                "PDP mean"

        )


        ax.set_xlabel(
            FEATURE_MAP[
                feature
            ]
        )


        ax.set_ylabel(
            f"Predicted {response_name} ({unit})"
        )


        ax.set_title(
            f"{response_name}: PDP + ICE"
        )


        ax.legend()


        fig.tight_layout()


        fig.savefig(

            PDP_DIR /
            f"{prefix}_{feature}_PDP_ICE.png",

            dpi=
                600,

            bbox_inches=
                "tight"

        )


        fig.savefig(

            PDP_DIR /
            f"{prefix}_{feature}_PDP_ICE.pdf",

            bbox_inches=
                "tight"

        )


        plt.show()


        pdp_summary.append({

            "Response":
                response_name,

            "Feature":
                feature,

            "PDP_Min":
                np.min(
                    pdp
                ),

            "PDP_Max":
                np.max(
                    pdp
                ),

            "PDP_Range":
                np.max(
                    pdp
                )
                -
                np.min(
                    pdp
                ),

            "Direction_Start_to_End":
                pdp[-1]
                -
                pdp[0]

        })


    return pd.DataFrame(
        pdp_summary
    )


pdp_t = run_pdp_ice(

    tensile_bundle,

    "Tensile Strength",

    "Tensile",

    "MPa"

)


pdp_r = run_pdp_ice(

    roughness_bundle,

    "Surface Roughness",

    "Roughness",

    "µm"

)


pdp_t.to_csv(

    TABLE_DIR /
    "Tensile_PDP_Summary.csv",

    index=False

)


pdp_r.to_csv(

    TABLE_DIR /
    "Roughness_PDP_Summary.csv",

    index=False

)


# =================================================================================================
# 18. TWO-WAY INTERACTION SURFACES
#
# Engineering-relevant interactions:
# GF x Layer
# GF x Nozzle Temperature
# Layer x Printing Speed
# =================================================================================================

INTERACTION_PAIRS = [

    (
        "GF_pct",
        "Layer_mm"
    ),

    (
        "GF_pct",
        "Nozzle_C"
    ),

    (
        "Layer_mm",
        "Speed_mm_s"
    )

]


interaction_bg_idx = rng.choice(

    len(
        X_ALL
    ),

    size=min(
        INTERACTION_BACKGROUND_N,
        len(
            X_ALL
        )
    ),

    replace=False

)


X_INTERACTION_BACKGROUND = (
    X_ALL[
        interaction_bg_idx
    ]
    .copy()
)


def interaction_surface(
    bundle,
    feature_a,
    feature_b
):

    ia = INPUT_COLS.index(
        feature_a
    )

    ib = INPUT_COLS.index(
        feature_b
    )


    grid_a = np.linspace(

        X_ALL[
            :,
            ia
        ].min(),

        X_ALL[
            :,
            ia
        ].max(),

        INTERACTION_GRID_N

    )


    grid_b = np.linspace(

        X_ALL[
            :,
            ib
        ].min(),

        X_ALL[
            :,
            ib
        ].max(),

        INTERACTION_GRID_N

    )


    surface = np.zeros(

        (
            len(
                grid_b
            ),
            len(
                grid_a
            )
        )

    )


    for i, value_b in enumerate(
        grid_b
    ):

        for j, value_a in enumerate(
            grid_a
        ):

            X_temp = (
                X_INTERACTION_BACKGROUND.copy()
            )


            X_temp[
                :,
                ia
            ] = value_a


            X_temp[
                :,
                ib
            ] = value_b


            surface[
                i,
                j
            ] = np.mean(

                predict_bundle(
                    bundle,
                    X_temp
                )

            )


    return (
        grid_a,
        grid_b,
        surface
    )


def run_interactions(
    bundle,
    response_name,
    prefix,
    unit
):

    print(
        f"\nRunning interaction surfaces — {response_name}"
    )


    for (
        feature_a,
        feature_b
    ) in INTERACTION_PAIRS:


        grid_a, grid_b, surface = (
            interaction_surface(

                bundle,

                feature_a,

                feature_b

            )
        )


        rows = []


        for i, b_value in enumerate(
            grid_b
        ):

            for j, a_value in enumerate(
                grid_a
            ):

                rows.append({

                    feature_a:
                        a_value,

                    feature_b:
                        b_value,

                    "Mean_Prediction":
                        surface[
                            i,
                            j
                        ]

                })


        interaction_df = pd.DataFrame(
            rows
        )


        interaction_df.to_csv(

            INTERACTION_DIR /
            f"{prefix}_{feature_a}_X_{feature_b}.csv",

            index=False

        )


        A, B = np.meshgrid(

            grid_a,

            grid_b

        )


        fig, ax = plt.subplots(

            figsize=(
                7,
                5.8
            )

        )


        contour = ax.contourf(

            A,

            B,

            surface,

            levels=
                20

        )


        cbar = fig.colorbar(

            contour,

            ax=ax

        )


        cbar.set_label(
            f"Predicted {response_name} ({unit})"
        )


        ax.set_xlabel(
            FEATURE_MAP[
                feature_a
            ]
        )


        ax.set_ylabel(
            FEATURE_MAP[
                feature_b
            ]
        )


        ax.set_title(
            f"{response_name}: "
            f"{FEATURE_MAP[feature_a]} × "
            f"{FEATURE_MAP[feature_b]}"
        )


        fig.tight_layout()


        safe_name = (
            f"{prefix}_{feature_a}_X_{feature_b}"
        )


        fig.savefig(

            INTERACTION_DIR /
            f"{safe_name}.png",

            dpi=
                600,

            bbox_inches=
                "tight"

        )


        fig.savefig(

            INTERACTION_DIR /
            f"{safe_name}.pdf",

            bbox_inches=
                "tight"

        )


        plt.show()


run_interactions(

    tensile_bundle,

    "Tensile Strength",

    "Tensile",

    "MPa"

)


run_interactions(

    roughness_bundle,

    "Surface Roughness",

    "Roughness",

    "µm"

)


# =================================================================================================
# 19. MASTER XAI TABLE
# =================================================================================================

master_xai = pd.concat(

    [
        combined_t,
        combined_r
    ],

    ignore_index=True

)


master_xai.to_csv(

    TABLE_DIR /
    "Cell5A_Master_XAI_Ranking.csv",

    index=False

)


# =================================================================================================
# 20. EXCEL WORKBOOK
# =================================================================================================

EXCEL_FILE = (

    OUTPUT_DIR /
    "Cell5A_Final_Model_Explainability.xlsx"

)


with pd.ExcelWriter(

    EXCEL_FILE,

    engine=
        "openpyxl"

) as writer:


    perm_t.to_excel(

        writer,

        sheet_name=
            "Tensile Permutation",

        index=False

    )


    perm_r.to_excel(

        writer,

        sheet_name=
            "Roughness Permutation",

        index=False

    )


    shap_t.to_excel(

        writer,

        sheet_name=
            "Tensile SHAP",

        index=False

    )


    shap_r.to_excel(

        writer,

        sheet_name=
            "Roughness SHAP",

        index=False

    )


    combined_t.to_excel(

        writer,

        sheet_name=
            "Tensile XAI Rank",

        index=False

    )


    combined_r.to_excel(

        writer,

        sheet_name=
            "Roughness XAI Rank",

        index=False

    )


    pdp_t.to_excel(

        writer,

        sheet_name=
            "Tensile PDP Summary",

        index=False

    )


    pdp_r.to_excel(

        writer,

        sheet_name=
            "Roughness PDP Summary",

        index=False

    )


# =================================================================================================
# 21. MANIFEST
# =================================================================================================

manifest5a = {

    "project_root":
        str(
            PROJECT_ROOT
        ),

    "output_dir":
        str(
            OUTPUT_DIR
        ),

    "data_file":
        str(
            DATA_FILE
        ),

    "tensile_final_model":
        str(
            TENSILE_MODEL_FILE
        ),

    "roughness_final_model":
        str(
            ROUGHNESS_MODEL_FILE
        ),

    "tensile_selected_model":
        tensile_bundle[
            "selected_model"
        ],

    "roughness_selected_model":
        roughness_bundle[
            "selected_model"
        ],

    "permutation_repeats":
        PERMUTATION_REPEATS,

    "shap_background_n":
        SHAP_BACKGROUND_N,

    "shap_explain_n":
        SHAP_EXPLAIN_N,

    "tensile_shap_method":
        shap_method_t,

    "roughness_shap_method":
        shap_method_r,

    "ice_sample_n":
        ICE_SAMPLE_N,

    "pdp_grid_n":
        PDP_GRID_N,

    "interaction_pairs":
        [
            list(
                pair
            )
            for pair
            in INTERACTION_PAIRS
        ],

    "excel_results":
        str(
            EXCEL_FILE
        ),

    "important_interpretation":
        (
            "Cell 5A explains the final surrogate models fitted to "
            "Cell 3 literature-informed virtual responses. "
            "Feature importance and SHAP values describe surrogate-model "
            "behavior and must not be interpreted as experimentally proven causality."
        ),

    "seed":
        SEED

}


MANIFEST_FILE = (

    OUTPUT_DIR /
    "Cell5A_Manifest.json"

)


with open(

    MANIFEST_FILE,

    "w",

    encoding=
        "utf-8"

) as f:

    json.dump(

        manifest5a,

        f,

        indent=
            4

    )


# =================================================================================================
# 22. FINAL SUMMARY
# =================================================================================================

print(
    "\n" +
    "=" * 110
)


print(
    "CELL 5A — FINAL SURROGATE EXPLAINABILITY COMPLETE"
)


print(
    "=" * 110
)


print(
    "\nTENSILE — TOP 3 GLOBAL XAI PARAMETERS"
)


for i, row in combined_t.head(
    3
).iterrows():

    print(

        f"{i+1}. "
        f"{row['Feature_Label']}"

    )


print(
    "\nROUGHNESS — TOP 3 GLOBAL XAI PARAMETERS"
)


for i, row in combined_r.head(
    3
).iterrows():

    print(

        f"{i+1}. "
        f"{row['Feature_Label']}"

    )


print(
    f"\nTensile SHAP method : "
    f"{shap_method_t}"
)


print(
    f"Roughness SHAP method: "
    f"{shap_method_r}"
)


print(
    f"\nExcel results:\n"
    f"{EXCEL_FILE}"
)


print(
    f"\nOutputs saved to:\n"
    f"{OUTPUT_DIR}"
)


print(
    "\nIMPORTANT:"
)


print(
    "These explanations describe the final virtual-response surrogate models."
)


print(
    "They do NOT establish experimental causality."
)


print(
    "\nNEXT: Cell 5B = Sobol global sensitivity analysis."
)


print(
    "=" * 110
)
```
