```python
# =================================================================================================
# CELL 2 — HIGH-QUALITY SIMULATION DESIGN
# Optimized Latin Hypercube Sampling + Scrambled Sobol Cross-Validation
# 500 / 1000 / 2000 / 4000 LHS Convergence
# Q1 Journal-Oriented | Python / Jupyter Notebook
# =================================================================================================

import sys
import subprocess
import importlib.util
import warnings
import json
import math
from pathlib import Path

warnings.filterwarnings("ignore")

# -------------------------------------------------------------------------------------------------
# 1. PACKAGE CHECK
# -------------------------------------------------------------------------------------------------

required_packages = {
    "numpy": "numpy",
    "pandas": "pandas",
    "scipy": "scipy>=1.10",
    "matplotlib": "matplotlib",
    "openpyxl": "openpyxl"
}

for module_name, package_name in required_packages.items():

    if importlib.util.find_spec(module_name) is None:

        subprocess.check_call(
            [
                sys.executable,
                "-m",
                "pip",
                "install",
                "-q",
                package_name
            ]
        )

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.stats import qmc, kstest
from scipy.spatial import cKDTree

from openpyxl import load_workbook
from openpyxl.styles import (
    Font,
    PatternFill,
    Alignment
)
from openpyxl.utils import get_column_letter

from IPython.display import display


# -------------------------------------------------------------------------------------------------
# 2. REPRODUCIBILITY
# -------------------------------------------------------------------------------------------------

SEED = 20260906

np.random.seed(SEED)

pd.set_option(
    "display.max_columns",
    100
)

pd.set_option(
    "display.width",
    200
)

pd.set_option(
    "display.float_format",
    lambda x: f"{x:.6f}"
)


# -------------------------------------------------------------------------------------------------
# 3. LOCATE CELL-1 OUTPUT
# -------------------------------------------------------------------------------------------------

candidate_projects = [

    Path(
        r"C:\Users\apurb\Downloads\Simulation OP ABS"
    ),

    Path.cwd(),

    Path.home() / "Downloads" / "Simulation OP ABS"
]


PROJECT_ROOT = None

CELL1_MANIFEST = None


for root in candidate_projects:

    manifest = (
        root /
        "Q1_ML_Output" /
        "Cell1_Data_Audit" /
        "Cell1_Manifest.json"
    )

    if manifest.exists():

        PROJECT_ROOT = root.resolve()

        CELL1_MANIFEST = manifest.resolve()

        break


if PROJECT_ROOT is None:

    raise FileNotFoundError(

        "\nCell 1 output was not found.\n\n"
        "Expected file:\n"
        "Q1_ML_Output/Cell1_Data_Audit/Cell1_Manifest.json\n\n"
        "Run Cell 1 first."
    )


with open(
    CELL1_MANIFEST,
    "r",
    encoding="utf-8"
) as f:

    cell1_manifest = json.load(f)


PARAMETER_BOUNDS_FILE = Path(
    cell1_manifest[
        "parameter_bounds"
    ]
)

DESIGN_POINTS_FILE = Path(
    cell1_manifest[
        "unique_design_points"
    ]
)


if not PARAMETER_BOUNDS_FILE.exists():

    raise FileNotFoundError(
        f"Parameter bounds file missing:\n{PARAMETER_BOUNDS_FILE}"
    )


# -------------------------------------------------------------------------------------------------
# 4. CELL-2 OUTPUT STRUCTURE
# -------------------------------------------------------------------------------------------------

OUTPUT_ROOT = (
    PROJECT_ROOT /
    "Q1_ML_Output"
)

CELL2_DIR = (
    OUTPUT_ROOT /
    "Cell2_Simulation_Design"
)

SIM_DATA_DIR = (
    CELL2_DIR /
    "Simulation_Data"
)

TABLE_DIR = (
    CELL2_DIR /
    "Tables"
)

FIGURE_DIR = (
    CELL2_DIR /
    "Figures"
)


for folder in [

    CELL2_DIR,
    SIM_DATA_DIR,
    TABLE_DIR,
    FIGURE_DIR

]:

    folder.mkdir(
        parents=True,
        exist_ok=True
    )


# -------------------------------------------------------------------------------------------------
# 5. LOAD REAL-DATA PARAMETER BOUNDS
# -------------------------------------------------------------------------------------------------

bounds_df = pd.read_csv(
    PARAMETER_BOUNDS_FILE
)


INPUT_COLS = [

    "GF_pct",
    "Infill_pct",
    "Layer_mm",
    "Nozzle_C",
    "Bed_C",
    "Speed_mm_s"

]


INPUT_LABELS = {

    "GF_pct":
        "Glass Fiber (%)",

    "Infill_pct":
        "Infill Density (%)",

    "Layer_mm":
        "Layer Thickness (mm)",

    "Nozzle_C":
        "Nozzle Temperature (°C)",

    "Bed_C":
        "Bed Temperature (°C)",

    "Speed_mm_s":
        "Printing Speed (mm/s)"
}


INPUT_UNITS = {

    "GF_pct": "%",

    "Infill_pct": "%",

    "Layer_mm": "mm",

    "Nozzle_C": "°C",

    "Bed_C": "°C",

    "Speed_mm_s": "mm/s"

}


required_columns = {

    "Variable",
    "Minimum",
    "Maximum"

}


if not required_columns.issubset(
    bounds_df.columns
):

    raise ValueError(
        "Cell 1 parameter-bound file has unexpected columns."
    )


bounds_df = (
    bounds_df
    .set_index("Variable")
    .loc[INPUT_COLS]
    .reset_index()
)


LOWER_BOUNDS = (
    bounds_df[
        "Minimum"
    ]
    .to_numpy(
        dtype=float
    )
)


UPPER_BOUNDS = (
    bounds_df[
        "Maximum"
    ]
    .to_numpy(
        dtype=float
    )
)


if np.any(
    UPPER_BOUNDS <= LOWER_BOUNDS
):

    raise ValueError(
        "One or more parameter bounds are invalid."
    )


DIMENSION = len(
    INPUT_COLS
)


# -------------------------------------------------------------------------------------------------
# 6. SAVE EXACT DESIGN-SPACE DEFINITION
# -------------------------------------------------------------------------------------------------

design_definition = pd.DataFrame({

    "Variable":
        INPUT_COLS,

    "Parameter":
        [
            INPUT_LABELS[x]
            for x in INPUT_COLS
        ],

    "Unit":
        [
            INPUT_UNITS[x]
            for x in INPUT_COLS
        ],

    "Lower_Bound":
        LOWER_BOUNDS,

    "Upper_Bound":
        UPPER_BOUNDS,

    "Range":
        UPPER_BOUNDS -
        LOWER_BOUNDS

})


design_definition.to_csv(

    TABLE_DIR /
    "Table_01_Simulation_Design_Space.csv",

    index=False
)


# -------------------------------------------------------------------------------------------------
# 7. SPACE-FILLING QUALITY METRICS
# -------------------------------------------------------------------------------------------------

def nearest_neighbor_statistics(
    sample
):

    tree = cKDTree(
        sample
    )

    try:

        distances, _ = tree.query(
            sample,
            k=2,
            workers=-1
        )

    except TypeError:

        distances, _ = tree.query(
            sample,
            k=2
        )

    nearest = distances[:, 1]

    return {

        "Min_NN_Distance":
            float(
                np.min(nearest)
            ),

        "Mean_NN_Distance":
            float(
                np.mean(nearest)
            ),

        "Median_NN_Distance":
            float(
                np.median(nearest)
            )

    }


def maximum_absolute_correlation(
    sample
):

    corr = np.corrcoef(
        sample,
        rowvar=False
    )

    mask = ~np.eye(
        corr.shape[0],
        dtype=bool
    )

    return float(
        np.max(
            np.abs(
                corr[mask]
            )
        )
    )


def marginal_uniformity(
    sample
):

    ks_values = []

    for j in range(
        sample.shape[1]
    ):

        stat = kstest(
            sample[:, j],
            "uniform"
        ).statistic

        ks_values.append(
            stat
        )

    return {

        "Mean_Marginal_KS":
            float(
                np.mean(
                    ks_values
                )
            ),

        "Max_Marginal_KS":
            float(
                np.max(
                    ks_values
                )
            )

    }


def calculate_design_metrics(
    sample
):

    metrics = {}

    metrics[
        "Centered_Discrepancy"
    ] = float(
        qmc.discrepancy(
            sample,
            method="CD"
        )
    )

    metrics[
        "Wraparound_Discrepancy"
    ] = float(
        qmc.discrepancy(
            sample,
            method="WD"
        )
    )

    metrics[
        "L2_Star_Discrepancy"
    ] = float(
        qmc.discrepancy(
            sample,
            method="L2-star"
        )
    )

    metrics[
        "Max_Abs_Correlation"
    ] = maximum_absolute_correlation(
        sample
    )

    metrics.update(
        nearest_neighbor_statistics(
            sample
        )
    )

    metrics.update(
        marginal_uniformity(
            sample
        )
    )

    return metrics


# -------------------------------------------------------------------------------------------------
# 8. HIGH-QUALITY OPTIMIZED LHS GENERATOR
#
# Multiple optimized LHS candidates are generated.
# The candidate with the LOWEST centered discrepancy is selected.
# -------------------------------------------------------------------------------------------------

def generate_best_optimized_lhs(
    n_samples,
    n_restarts,
    base_seed
):

    best_sample = None

    best_discrepancy = np.inf

    restart_records = []


    for restart in range(
        n_restarts
    ):

        current_seed = (
            base_seed +
            n_samples * 31 +
            restart * 1009
        )


        sampler = qmc.LatinHypercube(

            d=DIMENSION,

            scramble=True,

            optimization="random-cd",

            strength=1,

            seed=current_seed

        )


        unit_sample = sampler.random(
            n=n_samples
        )


        cd = qmc.discrepancy(

            unit_sample,

            method="CD"

        )


        restart_records.append({

            "Sample_Size":
                n_samples,

            "Restart":
                restart + 1,

            "Seed":
                current_seed,

            "Centered_Discrepancy":
                float(cd)

        })


        if cd < best_discrepancy:

            best_discrepancy = cd

            best_sample = (
                unit_sample.copy()
            )


    restart_df = pd.DataFrame(
        restart_records
    )


    return (
        best_sample,
        restart_df
    )


# -------------------------------------------------------------------------------------------------
# 9. LHS CONVERGENCE SAMPLE SIZES
# -------------------------------------------------------------------------------------------------

LHS_SAMPLE_SIZES = [

    500,
    1000,
    2000,
    4000

]


# More restarts for smaller designs, fewer for 4000 for computational efficiency
RESTART_MAP = {

    500: 5,

    1000: 5,

    2000: 4,

    4000: 3

}


lhs_unit_designs = {}

lhs_physical_designs = {}

lhs_metric_records = []

lhs_restart_tables = []


# -------------------------------------------------------------------------------------------------
# 10. GENERATE OPTIMIZED LHS DESIGNS
# -------------------------------------------------------------------------------------------------

print(
    "\nGenerating optimized Latin Hypercube designs..."
)


for n in LHS_SAMPLE_SIZES:

    print(
        f"  Optimizing LHS design: n = {n}"
    )


    unit_sample, restart_df = (
        generate_best_optimized_lhs(

            n_samples=n,

            n_restarts=
                RESTART_MAP[n],

            base_seed=SEED

        )
    )


    lhs_restart_tables.append(
        restart_df
    )


    physical_sample = qmc.scale(

        unit_sample,

        LOWER_BOUNDS,

        UPPER_BOUNDS

    )


    lhs_unit_designs[n] = (
        unit_sample
    )

    lhs_physical_designs[n] = (
        physical_sample
    )


    metrics = (
        calculate_design_metrics(
            unit_sample
        )
    )


    metrics[
        "Method"
    ] = "Optimized LHS"

    metrics[
        "Requested_N"
    ] = n

    metrics[
        "Actual_N"
    ] = n


    lhs_metric_records.append(
        metrics
    )


    design_df = pd.DataFrame(

        physical_sample,

        columns=INPUT_COLS

    )


    design_df.insert(

        0,

        "Simulation_Run_ID",

        [
            f"LHS{n}_{i:05d}"
            for i in range(
                1,
                n + 1
            )
        ]

    )


    design_df[
        "Sampling_Method"
    ] = "Optimized_LHS"


    design_df.to_csv(

        SIM_DATA_DIR /
        f"LHS_{n}_Simulation_Inputs.csv",

        index=False

    )


# -------------------------------------------------------------------------------------------------
# 11. SOBOL CROSS-CHECK
#
# Sobol balance is strongest for N = 2^m:
# 512 / 1024 / 2048 / 4096
# -------------------------------------------------------------------------------------------------

SOBOL_SAMPLE_SIZES = [

    512,
    1024,
    2048,
    4096

]


sobol_unit_designs = {}

sobol_physical_designs = {}

sobol_metric_records = []


print(
    "\nGenerating scrambled Sobol reference designs..."
)


for idx, n in enumerate(
    SOBOL_SAMPLE_SIZES
):

    print(
        f"  Generating Sobol design: n = {n}"
    )


    m = int(
        math.log2(n)
    )


    sampler = qmc.Sobol(

        d=DIMENSION,

        scramble=True,

        seed=
            SEED +
            50000 +
            idx * 101

    )


    unit_sample = (
        sampler.random_base2(
            m=m
        )
    )


    physical_sample = qmc.scale(

        unit_sample,

        LOWER_BOUNDS,

        UPPER_BOUNDS

    )


    sobol_unit_designs[n] = (
        unit_sample
    )

    sobol_physical_designs[n] = (
        physical_sample
    )


    metrics = (
        calculate_design_metrics(
            unit_sample
        )
    )


    metrics[
        "Method"
    ] = "Scrambled Sobol"

    metrics[
        "Requested_N"
    ] = n

    metrics[
        "Actual_N"
    ] = n


    sobol_metric_records.append(
        metrics
    )


    design_df = pd.DataFrame(

        physical_sample,

        columns=INPUT_COLS

    )


    design_df.insert(

        0,

        "Simulation_Run_ID",

        [
            f"SOBOL{n}_{i:05d}"
            for i in range(
                1,
                n + 1
            )
        ]

    )


    design_df[
        "Sampling_Method"
    ] = "Scrambled_Sobol"


    design_df.to_csv(

        SIM_DATA_DIR /
        f"Sobol_{n}_Reference_Inputs.csv",

        index=False

    )


# -------------------------------------------------------------------------------------------------
# 12. LHS QUALITY / CONVERGENCE TABLE
# -------------------------------------------------------------------------------------------------

lhs_metrics = pd.DataFrame(
    lhs_metric_records
)


metric_order = [

    "Method",

    "Requested_N",

    "Actual_N",

    "Centered_Discrepancy",

    "Wraparound_Discrepancy",

    "L2_Star_Discrepancy",

    "Max_Abs_Correlation",

    "Min_NN_Distance",

    "Mean_NN_Distance",

    "Median_NN_Distance",

    "Mean_Marginal_KS",

    "Max_Marginal_KS"

]


lhs_metrics = (
    lhs_metrics[
        metric_order
    ]
    .sort_values(
        "Actual_N"
    )
    .reset_index(
        drop=True
    )
)


lhs_metrics[
    "CD_Improvement_vs_Previous_pct"
] = np.nan


for i in range(
    1,
    len(lhs_metrics)
):

    previous = lhs_metrics.loc[
        i - 1,
        "Centered_Discrepancy"
    ]

    current = lhs_metrics.loc[
        i,
        "Centered_Discrepancy"
    ]

    improvement = (

        (previous - current) /
        previous *
        100

    )

    lhs_metrics.loc[
        i,
        "CD_Improvement_vs_Previous_pct"
    ] = improvement


# -------------------------------------------------------------------------------------------------
# 13. SOBOL QUALITY TABLE
# -------------------------------------------------------------------------------------------------

sobol_metrics = pd.DataFrame(
    sobol_metric_records
)


sobol_metrics = (
    sobol_metrics[
        metric_order
    ]
    .sort_values(
        "Actual_N"
    )
    .reset_index(
        drop=True
    )
)


# -------------------------------------------------------------------------------------------------
# 14. LHS vs SOBOL MATCHED-SCALE COMPARISON
# -------------------------------------------------------------------------------------------------

comparison_rows = []


for lhs_n, sobol_n in zip(

    LHS_SAMPLE_SIZES,

    SOBOL_SAMPLE_SIZES

):


    lhs_row = lhs_metrics[
        lhs_metrics[
            "Actual_N"
        ] == lhs_n
    ].iloc[0]


    sobol_row = sobol_metrics[
        sobol_metrics[
            "Actual_N"
        ] == sobol_n
    ].iloc[0]


    comparison_rows.append({

        "LHS_N":
            lhs_n,

        "Sobol_N":
            sobol_n,

        "LHS_Centered_Discrepancy":
            lhs_row[
                "Centered_Discrepancy"
            ],

        "Sobol_Centered_Discrepancy":
            sobol_row[
                "Centered_Discrepancy"
            ],

        "LHS_MaxAbsCorrelation":
            lhs_row[
                "Max_Abs_Correlation"
            ],

        "Sobol_MaxAbsCorrelation":
            sobol_row[
                "Max_Abs_Correlation"
            ],

        "LHS_MaxMarginalKS":
            lhs_row[
                "Max_Marginal_KS"
            ],

        "Sobol_MaxMarginalKS":
            sobol_row[
                "Max_Marginal_KS"
            ],

        "LHS_MeanNNDistance":
            lhs_row[
                "Mean_NN_Distance"
            ],

        "Sobol_MeanNNDistance":
            sobol_row[
                "Mean_NN_Distance"
            ]

    })


method_comparison = pd.DataFrame(
    comparison_rows
)


# -------------------------------------------------------------------------------------------------
# 15. SAVE ALL OPTIMIZATION RESTART RESULTS
# -------------------------------------------------------------------------------------------------

restart_results = pd.concat(

    lhs_restart_tables,

    ignore_index=True

)


restart_results.to_csv(

    TABLE_DIR /
    "Table_02_LHS_Optimization_Restarts.csv",

    index=False

)


# -------------------------------------------------------------------------------------------------
# 16. SAVE CONVERGENCE TABLES
# -------------------------------------------------------------------------------------------------

lhs_metrics.to_csv(

    TABLE_DIR /
    "Table_03_LHS_Convergence_Metrics.csv",

    index=False

)


sobol_metrics.to_csv(

    TABLE_DIR /
    "Table_04_Sobol_Reference_Metrics.csv",

    index=False

)


method_comparison.to_csv(

    TABLE_DIR /
    "Table_05_LHS_vs_Sobol_Comparison.csv",

    index=False

)


# -------------------------------------------------------------------------------------------------
# 17. FINAL SIMULATION INPUT DESIGN
#
# Use 4000 optimized LHS points as the master design.
# Cell 4 will later test whether ML performance already converges at 500/1000/2000.
# -------------------------------------------------------------------------------------------------

FINAL_N = 4000


final_unit_design = (
    lhs_unit_designs[
        FINAL_N
    ]
)


final_physical_design = (
    lhs_physical_designs[
        FINAL_N
    ]
)


final_design = pd.DataFrame(

    final_physical_design,

    columns=INPUT_COLS

)


final_design.insert(

    0,

    "Simulation_Run_ID",

    [
        f"SIM_{i:05d}"
        for i in range(
            1,
            FINAL_N + 1
        )
    ]

)


final_design[
    "Sampling_Method"
] = "Optimized_LHS"


# -------------------------------------------------------------------------------------------------
# 18. ENGINEERING-SAFE NUMERICAL PRECISION
#
# Keep the DOE continuous, but store sensible precision.
# No arbitrary discretization is imposed.
# -------------------------------------------------------------------------------------------------

final_design[
    "GF_pct"
] = final_design[
    "GF_pct"
].round(4)


final_design[
    "Infill_pct"
] = final_design[
    "Infill_pct"
].round(4)


final_design[
    "Layer_mm"
] = final_design[
    "Layer_mm"
].round(5)


final_design[
    "Nozzle_C"
] = final_design[
    "Nozzle_C"
].round(4)


final_design[
    "Bed_C"
] = final_design[
    "Bed_C"
].round(4)


final_design[
    "Speed_mm_s"
] = final_design[
    "Speed_mm_s"
].round(4)


# -------------------------------------------------------------------------------------------------
# 19. CHECK FINAL DESIGN
# -------------------------------------------------------------------------------------------------

duplicate_input_rows = int(

    final_design[
        INPUT_COLS
    ]
    .duplicated()
    .sum()

)


outside_bounds = {}


for i, col in enumerate(
    INPUT_COLS
):

    count = int(

        (
            (
                final_design[col] <
                LOWER_BOUNDS[i]
            )
            |
            (
                final_design[col] >
                UPPER_BOUNDS[i]
            )
        )
        .sum()

    )

    outside_bounds[
        col
    ] = count


if duplicate_input_rows > 0:

    raise ValueError(
        f"{duplicate_input_rows} duplicate simulation designs detected."
    )


if sum(
    outside_bounds.values()
) > 0:

    raise ValueError(
        "Final design contains values outside real-data bounds."
    )


# -------------------------------------------------------------------------------------------------
# 20. FINAL DESIGN RANGE VALIDATION TABLE
# -------------------------------------------------------------------------------------------------

validation_rows = []


for i, col in enumerate(
    INPUT_COLS
):

    validation_rows.append({

        "Variable":
            col,

        "Allowed_Min":
            LOWER_BOUNDS[i],

        "Generated_Min":
            final_design[
                col
            ].min(),

        "Generated_Max":
            final_design[
                col
            ].max(),

        "Allowed_Max":
            UPPER_BOUNDS[i],

        "Rows_Below_Min":
            int(
                (
                    final_design[col] <
                    LOWER_BOUNDS[i]
                ).sum()
            ),

        "Rows_Above_Max":
            int(
                (
                    final_design[col] >
                    UPPER_BOUNDS[i]
                ).sum()
            )

    })


range_validation = pd.DataFrame(
    validation_rows
)


range_validation.to_csv(

    TABLE_DIR /
    "Table_06_Final_Design_Range_Validation.csv",

    index=False

)


# -------------------------------------------------------------------------------------------------
# 21. SAVE FINAL 4000 INPUTS
# -------------------------------------------------------------------------------------------------

FINAL_CSV = (

    SIM_DATA_DIR /
    "FINAL_4000_Optimized_LHS_Simulation_Inputs.csv"

)


FINAL_XLSX = (

    SIM_DATA_DIR /
    "FINAL_4000_Optimized_LHS_Simulation_Inputs.xlsx"

)


final_design.to_csv(

    FINAL_CSV,

    index=False

)


final_design.to_excel(

    FINAL_XLSX,

    index=False,

    sheet_name="Simulation_Inputs"

)


# -------------------------------------------------------------------------------------------------
# 22. CREATE MASTER CELL-2 EXCEL WORKBOOK
# -------------------------------------------------------------------------------------------------

MASTER_EXCEL = (

    CELL2_DIR /
    "Cell2_Simulation_Design_Results.xlsx"

)


with pd.ExcelWriter(

    MASTER_EXCEL,

    engine="openpyxl"

) as writer:


    design_definition.to_excel(

        writer,

        sheet_name="Design_Space",

        index=False

    )


    lhs_metrics.to_excel(

        writer,

        sheet_name="LHS_Convergence",

        index=False

    )


    sobol_metrics.to_excel(

        writer,

        sheet_name="Sobol_Reference",

        index=False

    )


    method_comparison.to_excel(

        writer,

        sheet_name="Method_Comparison",

        index=False

    )


    range_validation.to_excel(

        writer,

        sheet_name="Range_Validation",

        index=False

    )


    restart_results.to_excel(

        writer,

        sheet_name="LHS_Restarts",

        index=False

    )


    final_design.to_excel(

        writer,

        sheet_name="Final_4000_LHS",

        index=False

    )


# -------------------------------------------------------------------------------------------------
# 23. PROFESSIONAL EXCEL FORMATTING
# -------------------------------------------------------------------------------------------------

wb = load_workbook(
    MASTER_EXCEL
)


header_fill = PatternFill(

    fill_type="solid",

    fgColor="1F4E78"

)


header_font = Font(

    name="Times New Roman",

    size=11,

    bold=True,

    color="FFFFFF"

)


body_font = Font(

    name="Times New Roman",

    size=10

)


for ws in wb.worksheets:

    ws.freeze_panes = "A2"

    ws.sheet_view.showGridLines = False

    ws.auto_filter.ref = (
        ws.dimensions
    )


    for cell in ws[1]:

        cell.fill = (
            header_fill
        )

        cell.font = (
            header_font
        )

        cell.alignment = Alignment(

            horizontal="center",

            vertical="center",

            wrap_text=True

        )


    for row in ws.iter_rows(
        min_row=2
    ):

        for cell in row:

            cell.font = (
                body_font
            )

            cell.alignment = Alignment(
                vertical="center"
            )


    for column_cells in ws.columns:

        max_length = 0

        column_letter = (
            get_column_letter(
                column_cells[
                    0
                ].column
            )
        )


        for cell in column_cells:

            try:

                max_length = max(

                    max_length,

                    len(
                        str(
                            cell.value
                        )
                    )

                )

            except Exception:

                pass


        ws.column_dimensions[
            column_letter
        ].width = min(

            max(
                max_length + 2,
                12
            ),

            28

        )


    ws.row_dimensions[
        1
    ].height = 30


wb.save(
    MASTER_EXCEL
)


# -------------------------------------------------------------------------------------------------
# 24. Q1 JOURNAL FIGURE SETTINGS
# -------------------------------------------------------------------------------------------------

plt.rcParams.update({

    "font.family":
        "serif",

    "font.size":
        10,

    "axes.labelsize":
        11,

    "xtick.labelsize":
        9,

    "ytick.labelsize":
        9,

    "legend.fontsize":
        9,

    "axes.linewidth":
        0.8,

    "pdf.fonttype":
        42,

    "ps.fonttype":
        42,

    "figure.facecolor":
        "white",

    "axes.facecolor":
        "white"

})


def polish_axis(
    ax
):

    ax.spines[
        "top"
    ].set_visible(
        False
    )

    ax.spines[
        "right"
    ].set_visible(
        False
    )

    ax.grid(

        linewidth=0.4,

        alpha=0.25

    )


def save_figure(
    fig,
    name
):

    fig.savefig(

        FIGURE_DIR /
        f"{name}_600dpi.png",

        dpi=600,

        bbox_inches="tight",

        facecolor="white"

    )


    fig.savefig(

        FIGURE_DIR /
        f"{name}.pdf",

        bbox_inches="tight",

        facecolor="white"

    )


    plt.show()

    plt.close(
        fig
    )


# -------------------------------------------------------------------------------------------------
# 25. FIGURE 1 — CENTERED DISCREPANCY CONVERGENCE
# -------------------------------------------------------------------------------------------------

fig, ax = plt.subplots(

    figsize=(
        6.6,
        4.6
    )

)


ax.plot(

    lhs_metrics[
        "Actual_N"
    ],

    lhs_metrics[
        "Centered_Discrepancy"
    ],

    marker="o",

    linewidth=1.5,

    markersize=6,

    label="Optimized LHS"

)


ax.plot(

    sobol_metrics[
        "Actual_N"
    ],

    sobol_metrics[
        "Centered_Discrepancy"
    ],

    marker="s",

    linestyle="--",

    linewidth=1.5,

    markersize=6,

    label="Scrambled Sobol"

)


ax.set_xlabel(
    "Number of simulation points"
)

ax.set_ylabel(
    "Centered discrepancy"
)

ax.set_yscale(
    "log"
)

ax.legend(
    frameon=False
)

polish_axis(
    ax
)

fig.tight_layout()

save_figure(

    fig,

    "Figure_01_Sampling_Discrepancy_Convergence"

)


# -------------------------------------------------------------------------------------------------
# 26. FIGURE 2 — PARAMETER CORRELATION CONVERGENCE
# -------------------------------------------------------------------------------------------------

fig, ax = plt.subplots(

    figsize=(
        6.6,
        4.6
    )

)


ax.plot(

    lhs_metrics[
        "Actual_N"
    ],

    lhs_metrics[
        "Max_Abs_Correlation"
    ],

    marker="o",

    linewidth=1.5,

    markersize=6,

    label="Optimized LHS"

)


ax.plot(

    sobol_metrics[
        "Actual_N"
    ],

    sobol_metrics[
        "Max_Abs_Correlation"
    ],

    marker="s",

    linestyle="--",

    linewidth=1.5,

    markersize=6,

    label="Scrambled Sobol"

)


ax.set_xlabel(
    "Number of simulation points"
)

ax.set_ylabel(
    "Maximum absolute inter-parameter correlation"
)

ax.legend(
    frameon=False
)

polish_axis(
    ax
)

fig.tight_layout()

save_figure(

    fig,

    "Figure_02_Interparameter_Correlation_Convergence"

)


# -------------------------------------------------------------------------------------------------
# 27. FIGURE 3 — MARGINAL UNIFORMITY CONVERGENCE
# -------------------------------------------------------------------------------------------------

fig, ax = plt.subplots(

    figsize=(
        6.6,
        4.6
    )

)


ax.plot(

    lhs_metrics[
        "Actual_N"
    ],

    lhs_metrics[
        "Max_Marginal_KS"
    ],

    marker="o",

    linewidth=1.5,

    markersize=6,

    label="Optimized LHS"

)


ax.plot(

    sobol_metrics[
        "Actual_N"
    ],

    sobol_metrics[
        "Max_Marginal_KS"
    ],

    marker="s",

    linestyle="--",

    linewidth=1.5,

    markersize=6,

    label="Scrambled Sobol"

)


ax.set_xlabel(
    "Number of simulation points"
)

ax.set_ylabel(
    "Maximum marginal KS statistic"
)

ax.legend(
    frameon=False
)

polish_axis(
    ax
)

fig.tight_layout()

save_figure(

    fig,

    "Figure_03_Marginal_Uniformity_Convergence"

)


# -------------------------------------------------------------------------------------------------
# 28. FIGURE 4 — FINAL LHS EMPIRICAL COVERAGE
# -------------------------------------------------------------------------------------------------

fig, ax = plt.subplots(

    figsize=(
        6.6,
        5.0
    )

)


for j, col in enumerate(
    INPUT_COLS
):

    values = np.sort(
        final_unit_design[
            :,
            j
        ]
    )

    empirical_cdf = (

        np.arange(
            1,
            len(values) + 1
        ) /
        len(values)

    )


    ax.plot(

        values,

        empirical_cdf,

        linewidth=1.1,

        alpha=0.90,

        label=INPUT_LABELS[
            col
        ]

    )


ax.plot(

    [0, 1],

    [0, 1],

    linestyle="--",

    linewidth=1.3,

    label="Ideal uniform coverage"

)


ax.set_xlabel(
    "Normalized parameter value"
)

ax.set_ylabel(
    "Empirical cumulative probability"
)

ax.set_xlim(
    0,
    1
)

ax.set_ylim(
    0,
    1
)

ax.legend(

    frameon=False,

    fontsize=8,

    loc="lower right"

)

polish_axis(
    ax
)

fig.tight_layout()

save_figure(

    fig,

    "Figure_04_Final_LHS_Marginal_Coverage"

)


# -------------------------------------------------------------------------------------------------
# 29. FIGURE 5 — FINAL LHS INTER-PARAMETER CORRELATION MATRIX
# -------------------------------------------------------------------------------------------------

final_corr = np.corrcoef(

    final_unit_design,

    rowvar=False

)


fig, ax = plt.subplots(

    figsize=(
        6.8,
        5.8
    )

)


image = ax.imshow(

    final_corr,

    vmin=-1,

    vmax=1

)


ax.set_xticks(

    range(
        DIMENSION
    )
)

ax.set_yticks(

    range(
        DIMENSION
    )
)


short_labels = [

    "GF",

    "Infill",

    "Layer",

    "Nozzle",

    "Bed",

    "Speed"

]


ax.set_xticklabels(

    short_labels,

    rotation=45,

    ha="right"

)


ax.set_yticklabels(
    short_labels
)


for i in range(
    DIMENSION
):

    for j in range(
        DIMENSION
    ):

        ax.text(

            j,

            i,

            f"{final_corr[i, j]:.3f}",

            ha="center",

            va="center",

            fontsize=8

        )


cbar = fig.colorbar(

    image,

    ax=ax,

    fraction=0.046,

    pad=0.04

)


cbar.set_label(
    "Pearson correlation coefficient"
)


fig.tight_layout()

save_figure(

    fig,

    "Figure_05_Final_LHS_Correlation_Matrix"

)


# -------------------------------------------------------------------------------------------------
# 30. SAVE CELL-2 MANIFEST
# -------------------------------------------------------------------------------------------------

CELL2_MANIFEST = (

    CELL2_DIR /
    "Cell2_Manifest.json"

)


manifest = {

    "project_root":
        str(
            PROJECT_ROOT
        ),

    "cell1_manifest":
        str(
            CELL1_MANIFEST
        ),

    "cell2_output":
        str(
            CELL2_DIR
        ),

    "parameter_bounds":
        str(
            PARAMETER_BOUNDS_FILE
        ),

    "final_simulation_inputs":
        str(
            FINAL_CSV
        ),

    "final_simulation_inputs_excel":
        str(
            FINAL_XLSX
        ),

    "master_results_excel":
        str(
            MASTER_EXCEL
        ),

    "sampling_method":
        "Optimized Latin Hypercube Sampling",

    "final_sample_size":
        FINAL_N,

    "lhs_convergence_sizes":
        LHS_SAMPLE_SIZES,

    "sobol_reference_sizes":
        SOBOL_SAMPLE_SIZES,

    "random_seed":
        SEED

}


with open(

    CELL2_MANIFEST,

    "w",

    encoding="utf-8"

) as f:

    json.dump(

        manifest,

        f,

        indent=4

    )


# -------------------------------------------------------------------------------------------------
# 31. DISPLAY Q1 TABLES
# -------------------------------------------------------------------------------------------------

print(
    "\nSIMULATION DESIGN SPACE"
)

display(
    design_definition.round(6)
)


print(
    "\nOPTIMIZED LHS CONVERGENCE"
)

display(
    lhs_metrics.round(8)
)


print(
    "\nSCRAMBLED SOBOL REFERENCE"
)

display(
    sobol_metrics.round(8)
)


print(
    "\nLHS vs SOBOL COMPARISON"
)

display(
    method_comparison.round(8)
)


print(
    "\nFINAL DESIGN RANGE VALIDATION"
)

display(
    range_validation.round(6)
)


# -------------------------------------------------------------------------------------------------
# 32. FINAL REPORT
# -------------------------------------------------------------------------------------------------

final_metrics = (

    lhs_metrics[

        lhs_metrics[
            "Actual_N"
        ] == FINAL_N

    ]
    .iloc[0]

)


print(
    "\n" +
    "=" * 110
)


print(
    "CELL 2 — OPTIMIZED SIMULATION DESIGN COMPLETE"
)


print(
    "=" * 110
)


print(
    f"\nProject root:\n{PROJECT_ROOT}"
)


print(
    "\nREAL-DATA-CONSTRAINED DESIGN SPACE"
)


for i, col in enumerate(
    INPUT_COLS
):

    print(

        f"{INPUT_LABELS[col]:<30} : "
        f"{LOWER_BOUNDS[i]:.5f} "
        f"to "
        f"{UPPER_BOUNDS[i]:.5f} "
        f"{INPUT_UNITS[col]}"

    )


print(
    "\nLHS CONVERGENCE DESIGNS GENERATED:"
)


for n in LHS_SAMPLE_SIZES:

    value = lhs_metrics.loc[

        lhs_metrics[
            "Actual_N"
        ] == n,

        "Centered_Discrepancy"

    ].iloc[0]


    print(

        f"{n:>5} runs"
        f" | Centered discrepancy = "
        f"{value:.10f}"

    )


print(
    "\nSOBOL CROSS-CHECK DESIGNS GENERATED:"
)


for n in SOBOL_SAMPLE_SIZES:

    value = sobol_metrics.loc[

        sobol_metrics[
            "Actual_N"
        ] == n,

        "Centered_Discrepancy"

    ].iloc[0]


    print(

        f"{n:>5} runs"
        f" | Centered discrepancy = "
        f"{value:.10f}"

    )


print(
    "\nFINAL MASTER SIMULATION DESIGN"
)


print(
    f"Sampling method              : Optimized Latin Hypercube"
)


print(
    f"Number of input variables    : {DIMENSION}"
)


print(
    f"Number of simulation points  : {FINAL_N}"
)


print(
    f"Duplicate input combinations : {duplicate_input_rows}"
)


print(
    f"Out-of-bound values          : {sum(outside_bounds.values())}"
)


print(
    f"Centered discrepancy         : "
    f"{final_metrics['Centered_Discrepancy']:.10f}"
)


print(
    f"Max absolute correlation     : "
    f"{final_metrics['Max_Abs_Correlation']:.8f}"
)


print(
    f"Max marginal KS statistic    : "
    f"{final_metrics['Max_Marginal_KS']:.8f}"
)


print(
    "\nIMPORTANT:"
)


print(
    "Cell 2 generated INPUT CONDITIONS ONLY."
)


print(
    "No Tensile Strength or Surface Roughness values were fabricated in this cell."
)


print(
    "\nFINAL 4000-RUN INPUT FILE:"
)


print(
    FINAL_CSV
)


print(
    "\nEXCEL VERSION:"
)


print(
    FINAL_XLSX
)


print(
    "\nALL CELL-2 OUTPUTS SAVED TO:"
)


print(
    CELL2_DIR
)


print(
    "\nFigures:"
)


print(
    "• 600-DPI PNG"
)


print(
    "• Vector PDF"
)


print(
    "\nNEXT STEP:"
)


print(
    "Cell 3 = scientifically constrained Tensile/Roughness response simulation + "
    "uncertainty + validation."
)


print(
    "=" * 110
)
```
