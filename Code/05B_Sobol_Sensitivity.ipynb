```python
# =================================================================================================
# CELL 5B — GLOBAL SENSITIVITY ANALYSIS (SOBOL)
# Tensile + Roughness
#
# Final surrogate model sensitivity
# Sobol S1, ST, S2 interaction analysis
# Consensus ranking with Cell 5A XAI
# Q1-style polished figures
# 600-DPI PNG + vector PDF
# =================================================================================================

import sys
import subprocess
import importlib.util
import warnings
import json
from pathlib import Path

warnings.filterwarnings("ignore")


# =================================================================================================
# 1. PACKAGE CHECK
# =================================================================================================

required = {
    "numpy": "numpy",
    "pandas": "pandas",
    "matplotlib": "matplotlib",
    "joblib": "joblib",
    "openpyxl": "openpyxl",
    "SALib": "SALib"
}

for module, package in required.items():

    if importlib.util.find_spec(module) is None:

        subprocess.check_call([
            sys.executable,
            "-m",
            "pip",
            "install",
            "-q",
            package
        ])


# =================================================================================================
# 2. IMPORTS
# =================================================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import joblib

from SALib.sample import sobol as sobol_sampler
from SALib.analyze import sobol as sobol_analyzer

from IPython.display import display


# =================================================================================================
# 3. SETTINGS
# =================================================================================================

SEED = 20260906

SOBOL_BASE_N = 1024
CALC_SECOND_ORDER = True

pd.set_option(
    "display.max_columns",
    100
)

pd.set_option(
    "display.width",
    240
)

pd.set_option(
    "display.float_format",
    lambda x: f"{x:.6f}"
)


plt.rcParams.update({
    "font.family": "serif",
    "font.size": 11,
    "axes.titlesize": 12,
    "axes.labelsize": 11,
    "xtick.labelsize": 10,
    "ytick.labelsize": 10,
    "figure.dpi": 150
})


# =================================================================================================
# 4. FIND PROJECT
# =================================================================================================

candidate_projects = [

    Path(
        r"C:\Users\apurb\Downloads\Simulation OP ABS"
    ),

    Path.cwd(),

    Path.home() /
    "Downloads" /
    "Simulation OP ABS"
]


PROJECT_ROOT = None


for root in candidate_projects:

    tensile_manifest_file = (
        root /
        "Q1_ML_Output" /
        "Cell4_ML_Benchmarking" /
        "Cell4B_Tensile_Deep_Tuning" /
        "Cell4B_Tensile_Manifest.json"
    )

    roughness_manifest_file = (
        root /
        "Q1_ML_Output" /
        "Cell4_ML_Benchmarking" /
        "Cell4B_Roughness_Deep_Tuning" /
        "Cell4B_Roughness_Manifest.json"
    )

    xai_master_file = (
        root /
        "Q1_ML_Output" /
        "Cell5_XAI_Sensitivity" /
        "Cell5A_Explainability" /
        "Tables" /
        "Cell5A_Master_XAI_Ranking.csv"
    )

    if tensile_manifest_file.exists() and roughness_manifest_file.exists():

        PROJECT_ROOT = root.resolve()
        TENSILE_MANIFEST_FILE = tensile_manifest_file.resolve()
        ROUGHNESS_MANIFEST_FILE = roughness_manifest_file.resolve()
        XAI_MASTER_FILE = xai_master_file.resolve()
        break


if PROJECT_ROOT is None:

    raise FileNotFoundError(
        "Cell 4B manifest files were not found."
    )


# =================================================================================================
# 5. LOAD MANIFESTS
# =================================================================================================

with open(
    TENSILE_MANIFEST_FILE,
    "r",
    encoding="utf-8"
) as f:

    tensile_manifest = json.load(f)


with open(
    ROUGHNESS_MANIFEST_FILE,
    "r",
    encoding="utf-8"
) as f:

    roughness_manifest = json.load(f)


# =================================================================================================
# 6. LOAD FINAL MODEL BUNDLES
# =================================================================================================

TENSILE_BUNDLE_FILE = Path(
    tensile_manifest[
        "final_model_bundle"
    ]
)

ROUGHNESS_BUNDLE_FILE = Path(
    roughness_manifest[
        "final_model_bundle"
    ]
)

if not TENSILE_BUNDLE_FILE.exists():
    raise FileNotFoundError(TENSILE_BUNDLE_FILE)

if not ROUGHNESS_BUNDLE_FILE.exists():
    raise FileNotFoundError(ROUGHNESS_BUNDLE_FILE)

tensile_bundle = joblib.load(
    TENSILE_BUNDLE_FILE
)

roughness_bundle = joblib.load(
    ROUGHNESS_BUNDLE_FILE
)


# =================================================================================================
# 7. LOAD DATA
# =================================================================================================

DATA_FILE = Path(
    tensile_manifest[
        "data_file"
    ]
)

data = pd.read_csv(
    DATA_FILE
)

INPUT_COLS = tensile_manifest[
    "input_cols"
]

TENSILE_TARGET = tensile_manifest[
    "target"
]

ROUGHNESS_TARGET = roughness_manifest[
    "target"
]


FEATURE_LABELS = [
    "Glass Fiber (%)",
    "Infill Density (%)",
    "Layer Thickness (mm)",
    "Nozzle Temperature (°C)",
    "Bed Temperature (°C)",
    "Printing Speed (mm/s)"
]

FEATURE_MAP = dict(
    zip(
        INPUT_COLS,
        FEATURE_LABELS
    )
)

X_ALL = data[
    INPUT_COLS
].to_numpy(
    dtype=float
)

Y_TENSILE = data[
    TENSILE_TARGET
].to_numpy(
    dtype=float
)

Y_ROUGHNESS = data[
    ROUGHNESS_TARGET
].to_numpy(
    dtype=float
)


# =================================================================================================
# 8. FINAL BUNDLE PREDICTION FUNCTION
# =================================================================================================

def predict_bundle(
    bundle,
    X
):

    X = np.asarray(
        X,
        dtype=float
    )

    selected = bundle[
        "selected_model"
    ]

    base_models = bundle[
        "base_models"
    ]

    base_order = bundle[
        "base_model_order"
    ]

    if selected in base_models:

        return np.asarray(
            base_models[
                selected
            ].predict(
                X
            )
        ).ravel()

    P = np.column_stack([
        base_models[name].predict(X)
        for name in base_order
    ])

    if selected == "Weighted_Ensemble":

        weights = np.asarray(
            bundle[
                "weighted_weights"
            ],
            dtype=float
        )

        return np.asarray(
            P @ weights
        ).ravel()

    if selected == "Ridge_Stacking":

        return np.asarray(
            bundle[
                "stacker"
            ].predict(
                P
            )
        ).ravel()

    raise ValueError(
        f"Unknown final model: {selected}"
    )


# =================================================================================================
# 9. OUTPUT DIRECTORIES
# =================================================================================================

OUTPUT_DIR = (
    PROJECT_ROOT /
    "Q1_ML_Output" /
    "Cell5_XAI_Sensitivity" /
    "Cell5B_Sobol_Sensitivity"
)

TABLE_DIR = OUTPUT_DIR / "Tables"
FIGURE_DIR = OUTPUT_DIR / "Figures"
RAW_DIR = OUTPUT_DIR / "Raw_Sobol_Output"

for folder in [
    OUTPUT_DIR,
    TABLE_DIR,
    FIGURE_DIR,
    RAW_DIR
]:
    folder.mkdir(
        parents=True,
        exist_ok=True
    )


# =================================================================================================
# 10. DEFINE SOBOL PROBLEM
# Use observed simulation-space bounds
# =================================================================================================

bounds = []

for col in INPUT_COLS:

    bounds.append([
        float(data[col].min()),
        float(data[col].max())
    ])


problem = {
    "num_vars": len(INPUT_COLS),
    "names": INPUT_COLS,
    "bounds": bounds
}


bounds_df = pd.DataFrame({
    "Feature": INPUT_COLS,
    "Feature_Label": FEATURE_LABELS,
    "Lower_Bound": [b[0] for b in bounds],
    "Upper_Bound": [b[1] for b in bounds]
})

bounds_df.to_csv(
    TABLE_DIR / "Sobol_Input_Bounds.csv",
    index=False
)


# =================================================================================================
# 11. GENERATE SOBOL SAMPLE
# =================================================================================================

print(
    "\nGenerating Sobol samples..."
)

X_sobol = sobol_sampler.sample(
    problem,
    N=SOBOL_BASE_N,
    calc_second_order=CALC_SECOND_ORDER,
    scramble=True,
    seed=SEED
)

X_sobol_df = pd.DataFrame(
    X_sobol,
    columns=INPUT_COLS
)

X_sobol_df.to_csv(
    RAW_DIR / "Sobol_Samples.csv",
    index=False
)

print(
    f"Sobol samples generated: {len(X_sobol_df)}"
)


# =================================================================================================
# 12. MODEL EVALUATION ON SOBOL SAMPLES
# =================================================================================================

print(
    "\nEvaluating final surrogate models on Sobol samples..."
)

Y_sobol_t = predict_bundle(
    tensile_bundle,
    X_sobol
)

Y_sobol_r = predict_bundle(
    roughness_bundle,
    X_sobol
)

pd.DataFrame({
    "Pred_Tensile_MPa": Y_sobol_t,
    "Pred_Roughness_um": Y_sobol_r
}).to_csv(
    RAW_DIR / "Sobol_Model_Predictions.csv",
    index=False
)


# =================================================================================================
# 13. SOBOL ANALYSIS
# =================================================================================================

print(
    "\nRunning Sobol analysis..."
)

Si_t = sobol_analyzer.analyze(
    problem,
    Y_sobol_t,
    calc_second_order=CALC_SECOND_ORDER,
    print_to_console=False
)

Si_r = sobol_analyzer.analyze(
    problem,
    Y_sobol_r,
    calc_second_order=CALC_SECOND_ORDER,
    print_to_console=False
)


# =================================================================================================
# 14. TABLE BUILDERS
# =================================================================================================

def sobol_main_table(
    Si,
    response_name
):

    df = pd.DataFrame({
        "Response": response_name,
        "Feature": INPUT_COLS,
        "Feature_Label": FEATURE_LABELS,
        "S1": Si["S1"],
        "S1_conf": Si["S1_conf"],
        "ST": Si["ST"],
        "ST_conf": Si["ST_conf"]
    })

    df["Interaction_Contribution"] = df["ST"] - df["S1"]

    df["ST_Rank"] = (
        df["ST"]
        .rank(
            ascending=False,
            method="dense"
        )
        .astype(int)
    )

    df["S1_Rank"] = (
        df["S1"]
        .rank(
            ascending=False,
            method="dense"
        )
        .astype(int)
    )

    df = df.sort_values(
        "ST",
        ascending=False
    ).reset_index(
        drop=True
    )

    return df


def sobol_s2_table(
    Si,
    response_name
):

    s2 = Si["S2"]
    s2_conf = Si["S2_conf"]

    rows = []

    for i in range(len(INPUT_COLS)):

        for j in range(i + 1, len(INPUT_COLS)):

            rows.append({
                "Response": response_name,
                "Feature_A": INPUT_COLS[i],
                "Feature_A_Label": FEATURE_MAP[INPUT_COLS[i]],
                "Feature_B": INPUT_COLS[j],
                "Feature_B_Label": FEATURE_MAP[INPUT_COLS[j]],
                "S2": s2[i, j],
                "S2_conf": s2_conf[i, j],
                "Abs_S2": abs(s2[i, j]) if pd.notna(s2[i, j]) else np.nan
            })

    df = pd.DataFrame(
        rows
    )

    df = df.sort_values(
        "Abs_S2",
        ascending=False
    ).reset_index(
        drop=True
    )

    df["S2_Rank"] = np.arange(
        1,
        len(df) + 1
    )

    return df


tensile_main = sobol_main_table(
    Si_t,
    "Tensile Strength"
)

roughness_main = sobol_main_table(
    Si_r,
    "Surface Roughness"
)

tensile_s2 = sobol_s2_table(
    Si_t,
    "Tensile Strength"
)

roughness_s2 = sobol_s2_table(
    Si_r,
    "Surface Roughness"
)

tensile_main.to_csv(
    TABLE_DIR / "Tensile_Sobol_Main_Indices.csv",
    index=False
)

roughness_main.to_csv(
    TABLE_DIR / "Roughness_Sobol_Main_Indices.csv",
    index=False
)

tensile_s2.to_csv(
    TABLE_DIR / "Tensile_Sobol_S2_Interactions.csv",
    index=False
)

roughness_s2.to_csv(
    TABLE_DIR / "Roughness_Sobol_S2_Interactions.csv",
    index=False
)

print(
    "\nTENSILE — SOBOL MAIN INDICES"
)
display(
    tensile_main.round(6)
)

print(
    "\nROUGHNESS — SOBOL MAIN INDICES"
)
display(
    roughness_main.round(6)
)


# =================================================================================================
# 15. PLOTTING FUNCTIONS
# =================================================================================================

def save_current_figure(
    folder,
    filename
):

    plt.tight_layout()
    plt.savefig(
        folder / f"{filename}.png",
        dpi=600,
        bbox_inches="tight"
    )
    plt.savefig(
        folder / f"{filename}.pdf",
        bbox_inches="tight"
    )
    plt.show()


def plot_total_order_bar(
    df,
    response_name,
    filename
):

    temp = df.sort_values(
        "ST",
        ascending=True
    )

    plt.figure(
        figsize=(8.2, 5.6)
    )

    plt.barh(
        temp["Feature_Label"],
        temp["ST"],
        xerr=temp["ST_conf"]
    )

    plt.xlabel(
        "Sobol Total-Order Index (ST)"
    )

    plt.ylabel(
        "Input Parameter"
    )

    plt.title(
        f"{response_name} — Global Total-Order Sensitivity"
    )

    save_current_figure(
        FIGURE_DIR,
        filename
    )


def plot_s1_st_comparison(
    df,
    response_name,
    filename
):

    temp = df.sort_values(
        "ST",
        ascending=True
    )

    y = np.arange(
        len(temp)
    )

    width = 0.36

    plt.figure(
        figsize=(8.4, 5.8)
    )

    plt.barh(
        y - width / 2,
        temp["S1"],
        height=width,
        xerr=temp["S1_conf"],
        label="First-order (S1)"
    )

    plt.barh(
        y + width / 2,
        temp["ST"],
        height=width,
        xerr=temp["ST_conf"],
        label="Total-order (ST)"
    )

    plt.yticks(
        y,
        temp["Feature_Label"]
    )

    plt.xlabel(
        "Sobol Sensitivity Index"
    )

    plt.ylabel(
        "Input Parameter"
    )

    plt.title(
        f"{response_name} — First-Order vs Total-Order Sensitivity"
    )

    plt.legend(
        frameon=False
    )

    save_current_figure(
        FIGURE_DIR,
        filename
    )


def plot_interaction_share(
    df,
    response_name,
    filename
):

    temp = df.sort_values(
        "Interaction_Contribution",
        ascending=True
    )

    plt.figure(
        figsize=(8.0, 5.6)
    )

    plt.barh(
        temp["Feature_Label"],
        temp["Interaction_Contribution"]
    )

    plt.xlabel(
        "Interaction Contribution (ST − S1)"
    )

    plt.ylabel(
        "Input Parameter"
    )

    plt.title(
        f"{response_name} — Interaction Contribution of Each Parameter"
    )

    save_current_figure(
        FIGURE_DIR,
        filename
    )


def plot_s2_heatmap(
    Si,
    response_name,
    filename
):

    mat = np.array(
        Si["S2"],
        dtype=float
    )

    masked = mat.copy()

    for i in range(masked.shape[0]):

        for j in range(masked.shape[1]):

            if j <= i:
                masked[i, j] = np.nan

    fig, ax = plt.subplots(
        figsize=(7.4, 6.4)
    )

    im = ax.imshow(
        masked,
        aspect="auto"
    )

    ax.set_xticks(
        np.arange(len(FEATURE_LABELS))
    )

    ax.set_yticks(
        np.arange(len(FEATURE_LABELS))
    )

    ax.set_xticklabels(
        FEATURE_LABELS,
        rotation=45,
        ha="right"
    )

    ax.set_yticklabels(
        FEATURE_LABELS
    )

    ax.set_title(
        f"{response_name} — Second-Order Interaction (S2)"
    )

    cbar = fig.colorbar(
        im,
        ax=ax
    )
    cbar.set_label(
        "Sobol S2 Index"
    )

    for i in range(len(FEATURE_LABELS)):

        for j in range(len(FEATURE_LABELS)):

            if np.isfinite(masked[i, j]):

                ax.text(
                    j,
                    i,
                    f"{masked[i, j]:.3f}",
                    ha="center",
                    va="center",
                    fontsize=8
                )

    fig.tight_layout()

    fig.savefig(
        FIGURE_DIR / f"{filename}.png",
        dpi=600,
        bbox_inches="tight"
    )

    fig.savefig(
        FIGURE_DIR / f"{filename}.pdf",
        bbox_inches="tight"
    )

    plt.show()


# =================================================================================================
# 16. GENERATE SOBOL FIGURES
# =================================================================================================

plot_total_order_bar(
    tensile_main,
    "Tensile Strength",
    "Tensile_Sobol_ST_Bar"
)

plot_total_order_bar(
    roughness_main,
    "Surface Roughness",
    "Roughness_Sobol_ST_Bar"
)

plot_s1_st_comparison(
    tensile_main,
    "Tensile Strength",
    "Tensile_Sobol_S1_vs_ST"
)

plot_s1_st_comparison(
    roughness_main,
    "Surface Roughness",
    "Roughness_Sobol_S1_vs_ST"
)

plot_interaction_share(
    tensile_main,
    "Tensile Strength",
    "Tensile_Interaction_Contribution"
)

plot_interaction_share(
    roughness_main,
    "Surface Roughness",
    "Roughness_Interaction_Contribution"
)

plot_s2_heatmap(
    Si_t,
    "Tensile Strength",
    "Tensile_Sobol_S2_Heatmap"
)

plot_s2_heatmap(
    Si_r,
    "Surface Roughness",
    "Roughness_Sobol_S2_Heatmap"
)


# =================================================================================================
# 17. TOP INTERACTION TABLES
# =================================================================================================

tensile_top_s2 = tensile_s2.head(10).copy()
roughness_top_s2 = roughness_s2.head(10).copy()

tensile_top_s2.to_csv(
    TABLE_DIR / "Tensile_Top10_S2_Interactions.csv",
    index=False
)

roughness_top_s2.to_csv(
    TABLE_DIR / "Roughness_Top10_S2_Interactions.csv",
    index=False
)


# =================================================================================================
# 18. CONSENSUS RANKING WITH CELL 5A XAI
# =================================================================================================

consensus_tables = []

if XAI_MASTER_FILE.exists():

    xai_master = pd.read_csv(
        XAI_MASTER_FILE
    )

    response_map = {
        "Tensile Strength": tensile_main,
        "Surface Roughness": roughness_main
    }

    for response_name, sobol_df in response_map.items():

        xai_df = xai_master[
            xai_master["Response"] == response_name
        ].copy()

        merged = xai_df.merge(
            sobol_df[[
                "Feature",
                "Feature_Label",
                "ST",
                "ST_conf",
                "ST_Rank",
                "S1",
                "S1_conf",
                "Interaction_Contribution"
            ]],
            on=["Feature", "Feature_Label"],
            how="left"
        )

        merged["Consensus_Mean_Rank"] = (
            merged["Combined_Rank"] + merged["ST_Rank"]
        ) / 2

        merged = merged.sort_values(
            "Consensus_Mean_Rank",
            ascending=True
        ).reset_index(
            drop=True
        )

        merged["Consensus_Rank"] = np.arange(
            1,
            len(merged) + 1
        )

        consensus_tables.append(
            merged
        )

    consensus_all = pd.concat(
        consensus_tables,
        ignore_index=True
    )

    consensus_all.to_csv(
        TABLE_DIR / "Cell5B_Consensus_Ranking_XAI_Sobol.csv",
        index=False
    )

    print(
        "\nCONSENSUS RANKING — XAI + SOBOL"
    )
    display(
        consensus_all.round(6)
    )


    def plot_consensus(
        df,
        response_name,
        filename
    ):

        temp = df[
            df["Response"] == response_name
        ].copy()

        temp = temp.sort_values(
            "Consensus_Mean_Rank",
            ascending=False
        )

        plt.figure(
            figsize=(8.1, 5.8)
        )

        plt.barh(
            temp["Feature_Label"],
            temp["Consensus_Mean_Rank"]
        )

        plt.xlabel(
            "Mean Rank (Lower = More Important)"
        )

        plt.ylabel(
            "Input Parameter"
        )

        plt.title(
            f"{response_name} — Consensus Importance Ranking (XAI + Sobol)"
        )

        save_current_figure(
            FIGURE_DIR,
            filename
        )

    plot_consensus(
        consensus_all,
        "Tensile Strength",
        "Tensile_Consensus_Ranking"
    )

    plot_consensus(
        consensus_all,
        "Surface Roughness",
        "Roughness_Consensus_Ranking"
    )

else:

    consensus_all = pd.DataFrame()
    print(
        "\nCell 5A master XAI ranking not found. Consensus ranking skipped."
    )


# =================================================================================================
# 19. EXCEL WORKBOOK
# =================================================================================================

EXCEL_FILE = OUTPUT_DIR / "Cell5B_Sobol_Sensitivity_Results.xlsx"

with pd.ExcelWriter(
    EXCEL_FILE,
    engine="openpyxl"
) as writer:

    bounds_df.to_excel(
        writer,
        sheet_name="Input Bounds",
        index=False
    )

    tensile_main.to_excel(
        writer,
        sheet_name="Tensile Main",
        index=False
    )

    roughness_main.to_excel(
        writer,
        sheet_name="Roughness Main",
        index=False
    )

    tensile_s2.to_excel(
        writer,
        sheet_name="Tensile S2",
        index=False
    )

    roughness_s2.to_excel(
        writer,
        sheet_name="Roughness S2",
        index=False
    )

    tensile_top_s2.to_excel(
        writer,
        sheet_name="Tensile Top10 S2",
        index=False
    )

    roughness_top_s2.to_excel(
        writer,
        sheet_name="Roughness Top10 S2",
        index=False
    )

    if len(consensus_all) > 0:

        consensus_all.to_excel(
            writer,
            sheet_name="Consensus XAI Sobol",
            index=False
        )


# =================================================================================================
# 20. MANIFEST
# =================================================================================================

manifest5b = {
    "project_root": str(PROJECT_ROOT),
    "output_dir": str(OUTPUT_DIR),
    "excel_results": str(EXCEL_FILE),
    "data_file": str(DATA_FILE),
    "tensile_model_bundle": str(TENSILE_BUNDLE_FILE),
    "roughness_model_bundle": str(ROUGHNESS_BUNDLE_FILE),
    "sobol_base_N": SOBOL_BASE_N,
    "calc_second_order": CALC_SECOND_ORDER,
    "sample_count": int(len(X_sobol_df)),
    "input_cols": INPUT_COLS,
    "feature_labels": FEATURE_LABELS,
    "important_note": (
        "Sobol sensitivity indices describe the final surrogate response surfaces "
        "trained on literature-informed virtual data. These are surrogate-level "
        "global sensitivities and should not be overclaimed as direct experimental causality."
    ),
    "seed": SEED
}

MANIFEST_FILE = OUTPUT_DIR / "Cell5B_Manifest.json"

with open(
    MANIFEST_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest5b,
        f,
        indent=4
    )


# =================================================================================================
# 21. FINAL SUMMARY
# =================================================================================================

print(
    "\n" + "=" * 110
)
print(
    "CELL 5B — SOBOL GLOBAL SENSITIVITY ANALYSIS COMPLETE"
)
print(
    "=" * 110
)

print(
    "\nTENSILE — TOP 3 PARAMETERS BY TOTAL-ORDER SENSITIVITY"
)
for i, row in tensile_main.head(3).iterrows():
    print(
        f"{i+1}. {row['Feature_Label']} | ST = {row['ST']:.6f}"
    )

print(
    "\nROUGHNESS — TOP 3 PARAMETERS BY TOTAL-ORDER SENSITIVITY"
)
for i, row in roughness_main.head(3).iterrows():
    print(
        f"{i+1}. {row['Feature_Label']} | ST = {row['ST']:.6f}"
    )

print(
    "\nTENSILE — TOP 5 SECOND-ORDER INTERACTIONS"
)
for i, row in tensile_s2.head(5).iterrows():
    print(
        f"{i+1}. {row['Feature_A_Label']} × {row['Feature_B_Label']} | S2 = {row['S2']:.6f}"
    )

print(
    "\nROUGHNESS — TOP 5 SECOND-ORDER INTERACTIONS"
)
for i, row in roughness_s2.head(5).iterrows():
    print(
        f"{i+1}. {row['Feature_A_Label']} × {row['Feature_B_Label']} | S2 = {row['S2']:.6f}"
    )

print(
    f"\nExcel results:\n{EXCEL_FILE}"
)

print(
    f"\nOutputs saved to:\n{OUTPUT_DIR}"
)

print(
    "\nIMPORTANT:"
)
print(
    "Sobol indices quantify the behavior of the final virtual-response surrogate models."
)
print(
    "They should be reported as global sensitivity of the surrogate response surface,"
)
print(
    "not as direct proof of experimental causality."
)

print(
    "\nNEXT: Cell 6 = Multi-objective optimization (NSGA-II / Pareto front / TOPSIS)."
)

print(
    "=" * 110
)
```
